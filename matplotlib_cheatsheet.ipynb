# Matplotlib Cheat Sheet
A comprehensive, fully-executed reference for core Matplotlib plotting — line/bar/scatter plots, styling, subplots, annotations, and more. Every line has a comment explaining its use.

**Run all cells top to bottom** (or just read the outputs — they're already executed).

```python
import matplotlib  # import the matplotlib base package
import matplotlib.pyplot as plt  # import the pyplot plotting interface
import numpy as np  # import numpy for generating sample data
%matplotlib inline
matplotlib.__version__  # check installed matplotlib version
```

## 1. Basic Line Plot

```python
x = np.linspace(0, 10, 100)  # 100 evenly spaced x values from 0 to 10
y = np.sin(x)  # compute sine of each x value

plt.plot(x, y)  # draw a line plot of y vs x
plt.title("Basic Line Plot")  # set the plot title
plt.xlabel("x")  # label the x-axis
plt.ylabel("sin(x)")  # label the y-axis
plt.show()  # render the figure
```

## 2. Multiple Lines & Legends

```python
x = np.linspace(0, 10, 100)  # shared x values
y1 = np.sin(x)  # first curve: sine
y2 = np.cos(x)  # second curve: cosine

plt.plot(x, y1, label="sin(x)")  # plot first curve with a label
plt.plot(x, y2, label="cos(x)")  # plot second curve with a label
plt.legend()  # show legend using the labels above
plt.title("Multiple Lines")  # set the plot title
plt.show()  # render the figure
```

## 3. Line Styles, Colors & Markers

```python
x = np.arange(0, 10)  # simple integer x values

plt.plot(x, x, color="blue", linestyle="-", label="solid")  # solid blue line
plt.plot(x, x + 2, color="red", linestyle="--", label="dashed")  # dashed red line
plt.plot(x, x + 4, color="green", linestyle=":", marker="o", label="dotted+marker")  # dotted line with circle markers
plt.plot(x, x + 6, "k^-", label="shorthand style")  # shorthand string: black, triangle marker, solid line
plt.legend()  # show legend
plt.title("Line Styles, Colors & Markers")  # set the plot title
plt.show()  # render the figure
```

## 4. Scatter Plot

```python
rng = np.random.default_rng(0)  # seeded random generator for reproducibility
x = rng.random(50)  # 50 random x values
y = rng.random(50)  # 50 random y values
sizes = rng.integers(20, 200, size=50)  # random marker sizes
colors = rng.random(50)  # random values used to color-map the points

plt.scatter(x, y, s=sizes, c=colors, cmap="viridis", alpha=0.7)  # scatter plot with size/color/transparency mapping
plt.colorbar(label="color value")  # show a colorbar for the color mapping
plt.title("Scatter Plot")  # set the plot title
plt.show()  # render the figure
```

## 5. Bar Chart

```python
categories = ["A", "B", "C", "D"]  # category labels
values = [23, 45, 12, 38]  # values for each category

plt.bar(categories, values, color="skyblue")  # vertical bar chart
plt.title("Bar Chart")  # set the plot title
plt.ylabel("Value")  # label the y-axis
plt.show()  # render the figure
```

```python
categories = ["A", "B", "C", "D"]  # category labels
values = [23, 45, 12, 38]  # values for each category

plt.barh(categories, values, color="salmon")  # horizontal bar chart
plt.title("Horizontal Bar Chart")  # set the plot title
plt.xlabel("Value")  # label the x-axis
plt.show()  # render the figure
```

```python
labels = ["Q1", "Q2", "Q3"]  # x-axis group labels
men = [20, 34, 30]  # first group's values
women = [25, 32, 34]  # second group's values
x = np.arange(len(labels))  # numeric positions for the groups
width = 0.35  # width of each bar

plt.bar(x - width/2, men, width, label="Men")  # first set of bars, shifted left
plt.bar(x + width/2, women, width, label="Women")  # second set of bars, shifted right
plt.xticks(x, labels)  # set x-tick positions and labels
plt.legend()  # show legend
plt.title("Grouped Bar Chart")  # set the plot title
plt.show()  # render the figure
```

## 6. Histogram

```python
rng = np.random.default_rng(1)  # seeded random generator
data = rng.normal(0, 1, 1000)  # 1000 samples from a normal distribution

plt.hist(data, bins=30, color="purple", alpha=0.7, edgecolor="black")  # histogram with 30 bins
plt.title("Histogram")  # set the plot title
plt.xlabel("Value")  # label the x-axis
plt.ylabel("Frequency")  # label the y-axis
plt.show()  # render the figure
```

## 7. Pie Chart

```python
labels = ["Python", "R", "SQL", "Other"]  # slice labels
sizes = [45, 20, 25, 10]  # slice sizes (percentages)
explode = (0.1, 0, 0, 0)  # slightly separate the first slice

plt.pie(sizes, labels=labels, explode=explode, autopct="%1.1f%%", startangle=90)  # pie chart with percentage labels
plt.title("Pie Chart")  # set the plot title
plt.axis("equal")  # keep pie chart circular, not elliptical
plt.show()  # render the figure
```

## 8. Subplots

```python
x = np.linspace(0, 10, 100)  # shared x values

fig, axes = plt.subplots(2, 2, figsize=(8, 6))  # create a 2x2 grid of subplots

axes[0, 0].plot(x, np.sin(x))  # top-left: sine curve
axes[0, 0].set_title("sin(x)")  # title for top-left subplot

axes[0, 1].plot(x, np.cos(x), color="orange")  # top-right: cosine curve
axes[0, 1].set_title("cos(x)")  # title for top-right subplot

axes[1, 0].plot(x, np.tan(x), color="green")  # bottom-left: tangent curve
axes[1, 0].set_ylim(-5, 5)  # limit y-axis since tan diverges
axes[1, 0].set_title("tan(x)")  # title for bottom-left subplot

axes[1, 1].plot(x, x**2, color="red")  # bottom-right: quadratic curve
axes[1, 1].set_title("x^2")  # title for bottom-right subplot

fig.tight_layout()  # auto-adjust spacing to prevent overlap
plt.show()  # render the figure
```

## 9. Figure Size & DPI

```python
x = np.linspace(0, 10, 100)  # shared x values
y = np.sin(x)  # sine values

plt.figure(figsize=(8, 3), dpi=100)  # create a figure with custom width/height (inches) and resolution
plt.plot(x, y)  # plot the curve
plt.title("Custom Figure Size")  # set the plot title
plt.show()  # render the figure
```

## 10. Axis Limits, Ticks & Grid

```python
x = np.linspace(0, 10, 100)  # shared x values
y = np.sin(x)  # sine values

plt.plot(x, y)  # plot the curve
plt.xlim(0, 8)  # restrict x-axis range
plt.ylim(-1.5, 1.5)  # restrict y-axis range
plt.xticks(np.arange(0, 9, 2))  # set custom x-axis tick positions
plt.grid(True, linestyle="--", alpha=0.5)  # show a dashed, semi-transparent grid
plt.title("Axis Limits, Ticks & Grid")  # set the plot title
plt.show()  # render the figure
```

## 11. Annotations & Text

```python
x = np.linspace(0, 10, 100)  # shared x values
y = np.sin(x)  # sine values

plt.plot(x, y)  # plot the curve
plt.annotate("max", xy=(np.pi/2, 1), xytext=(3, 1.3),  # annotate the peak with an arrow
             arrowprops=dict(facecolor="black", shrink=0.05))  # arrow style pointing to the peak
plt.text(6, -1.2, "sample text")  # place plain text at a specific coordinate
plt.title("Annotations & Text")  # set the plot title
plt.show()  # render the figure
```

## 12. Filling Areas Under Curves

```python
x = np.linspace(0, 10, 100)  # shared x values
y = np.sin(x)  # sine values

plt.plot(x, y, color="blue")  # plot the curve
plt.fill_between(x, y, 0, where=(y > 0), color="blue", alpha=0.3)  # shade area where curve is positive
plt.fill_between(x, y, 0, where=(y <= 0), color="red", alpha=0.3)  # shade area where curve is negative
plt.title("Filled Area Under Curve")  # set the plot title
plt.show()  # render the figure
```

## 13. Error Bars

```python
x = np.arange(5)  # x positions
y = np.array([2, 4, 3, 5, 4])  # y values
error = np.array([0.3, 0.5, 0.2, 0.4, 0.3])  # error magnitude for each point

plt.errorbar(x, y, yerr=error, fmt="o-", capsize=5, color="darkgreen")  # line plot with vertical error bars
plt.title("Error Bars")  # set the plot title
plt.show()  # render the figure
```

## 14. Box Plot

```python
rng = np.random.default_rng(2)  # seeded random generator
data = [rng.normal(0, std, 100) for std in range(1, 4)]  # three datasets with increasing variance

plt.boxplot(data, labels=["std=1", "std=2", "std=3"])  # box plot summarizing distribution per dataset
plt.title("Box Plot")  # set the plot title
plt.show()  # render the figure
```

## 15. Log Scale

```python
x = np.linspace(1, 100, 100)  # x values from 1 to 100
y = x ** 2  # quadratic growth

plt.plot(x, y)  # plot the curve
plt.yscale("log")  # set y-axis to logarithmic scale
plt.title("Log Scale (y-axis)")  # set the plot title
plt.show()  # render the figure
```

## 16. Saving Figures

```python
x = np.linspace(0, 10, 100)  # shared x values
y = np.sin(x)  # sine values

plt.plot(x, y)  # plot the curve
plt.title("Figure to Save")  # set the plot title
plt.savefig("/tmp/my_plot.png", dpi=150, bbox_inches="tight")  # save figure as PNG, tightly cropped
plt.show()  # render the figure (also visible after saving)
```

## 17. Styles & Themes

```python
print(plt.style.available[:8])  # list some of the built-in style names available

x = np.linspace(0, 10, 100)  # shared x values
y = np.sin(x)  # sine values

with plt.style.context("ggplot"):  # temporarily apply the 'ggplot' style
    plt.plot(x, y)  # plot the curve using that style
    plt.title("ggplot Style")  # set the plot title
    plt.show()  # render the figure
```

## 18. Object-Oriented Interface (fig, ax)

```python
x = np.linspace(0, 10, 100)  # shared x values
y = np.sin(x)  # sine values

fig, ax = plt.subplots(figsize=(6, 4))  # create a figure and a single axes object
ax.plot(x, y, label="sin(x)")  # plot on the axes object directly
ax.set_xlabel("x")  # set x-axis label via the axes object
ax.set_ylabel("y")  # set y-axis label via the axes object
ax.set_title("Object-Oriented API")  # set title via the axes object
ax.legend()  # show legend via the axes object
plt.show()  # render the figure
```

## 19. Quick Reference Table

| Task | Function |
|---|---|
| Line plot | `plt.plot()` |
| Scatter plot | `plt.scatter()` |
| Bar chart | `plt.bar()`, `plt.barh()` |
| Histogram | `plt.hist()` |
| Pie chart | `plt.pie()` |
| Box plot | `plt.boxplot()` |
| Error bars | `plt.errorbar()` |
| Subplots | `plt.subplots(rows, cols)` |
| Labels/Title | `plt.xlabel()`, `plt.ylabel()`, `plt.title()` |
| Legend | `plt.legend()` |
| Axis limits | `plt.xlim()`, `plt.ylim()` |
| Grid | `plt.grid(True)` |
| Annotations | `plt.annotate()`, `plt.text()` |
| Fill area | `plt.fill_between()` |
| Log scale | `plt.xscale('log')`, `plt.yscale('log')` |
| Save figure | `plt.savefig()` |
| Styles | `plt.style.use()`, `plt.style.context()` |
| OO interface | `fig, ax = plt.subplots()` then `ax.plot()` |

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

df = pd.read_csv("C:/Users/This_PC/Desktop/Python/Python-Matplotlib/Data/ecommerce_orders.csv")

df.head(20)
```

```python
category_count = df["category"].value_counts()

plt.bar(category_count.index, category_count.values, color="skyblue",)

plt.title("Categories", fontweight="bold", fontsize=15)
plt.show()
```

```python
areas = df["region"].value_counts()
x = areas.index
y = areas.values
plt.grid(axis="y", linestyle="dotted")
plt.plot(x, y, marker="o", markerfacecolor="orange", markersize= 7)

plt.title("AREAS", fontsize=15, color="purple", fontweight="bold")

plt.show()
```

```python
payment = df["payment_method"].value_counts()
method = payment.index
values = payment.values
# print(values)
# print(method)
plt.pie(values, labels=method, autopct="%1.1f%%")
plt.title("PAYEMENT METHOD", fontsize=15, fontweight="bold")
plt.show()
```

```python
products = df["product"].value_counts()
products_name = products.index
products_sold = products.values

plt.barh(products_name, products_sold)
plt.title("No Products Sold")
plt.show()
```
